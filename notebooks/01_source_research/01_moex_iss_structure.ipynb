## Получение истории торгов

Цель:
Практический кейс для понимания работы MOEX ISS

Источник:
https://moexapi.tech-order.ru/common_info_about_api

Входные условия:
Инструмент, рынок, режим торгов, период.


```python
import requests
import pandas as pd

pd.set_option('display.max_columns', None)
```

```python
#Получение информации о доступных торговых системах

engine_url = 'https://iss.moex.com/iss/engines.json'

engine_response = requests.get(
    url=engine_url,
    timeout=30
)
print(engine_response.status_code)

engines_data = engine_response.json()
engines_df = pd.DataFrame(
    data=engines_data["engines"]["data"],
    columns=engines_data["engines"]["columns"],
)

engines_df.head()
```

```python
#Получение информации о доступных рынках в системе "stock" - Фондовый рынок и рынок депозитов

markets_engines = 'stock'
markets_url = rf'https://iss.moex.com/iss/engines/{markets_engines}/markets.json'

markets_response = requests.get(
    url=markets_url,
    timeout=30
)
print(markets_response.status_code)

markets_data = markets_response.json()
markets_df = pd.DataFrame(
    data=markets_data['markets']['data'],
    columns=markets_data['markets']['columns']
)

markets_df.head(15)
```

```python
#Получение информации о режиме торгов

board_engines = 'stock'
board_markets = 'shares'
board_url = rf'https://iss.moex.com/iss/engines/{board_engines}/markets/{board_markets}/board.json'

board_response = requests.get(
    url=board_url,
    timeout=30
)
print(board_url)
print(board_response.status_code)

board_data = board_response.json()
board_df = pd.DataFrame(
    data=board_data['boards']['data'],
    columns=board_data['boards']['columns']
)

board_df.head(15)
```

```python
#Список инструментов выбранного режима
securities_engines = 'stock'
securities_markets = 'index'
securities_board = 'SNDX'
securities_url = rf'https://iss.moex.com/iss/engines/{securities_engines}/markets/{securities_markets}/boards/{securities_board}/securities.json'

securities_params = {
    "iss.only": "securities",
}

securities_response = requests.get(
    url=securities_url,
    params=securities_params,
    timeout=30
)

print(securities_url)
print(securities_response.status_code)

securities_data = securities_response.json()
securities_df = pd.DataFrame(
    data=securities_data['securities']['data'],
    columns=securities_data['securities']['columns']
)

securities_df.head(5)
```

```python
#Дневная история IMOEX за период
history_engine = "stock"
history_market = "index"
history_board = "SNDX"
history_security = "IMOEX"

history_url = rf'https://iss.moex.com/iss/history/engines/{history_engine}/markets/{history_market}/boards/{history_board}/securities/{history_security}.json'
history_params = {
    "from": '2026-07-01',
    "till": '2026-07-15',
    "lang": 'ru',
}

history_response = requests.get(
    url=history_url,
    params=history_params,
    timeout=30
)

print(history_url)
print(history_response.status_code)

history_data = history_response.json()


history_df = pd.DataFrame(
    data=history_data["history"]["data"],
    columns=history_data["history"]["columns"],
)

history_df.head()
```

```python
#Получить свечи указанного инструмента по выбранному режиму торгов IMOEX - 1час
candles_engine = "stock"
candles_market = "index"
candles_board = "SNDX"
candles_security = "IMOEX"

candles_url = rf'https://iss.moex.com/iss/engines/{candles_engine}/markets/{candles_market}/boards/{candles_board}/securities/{candles_security}/candles.json'
candles_params = {
    "from": '2026-07-01',
    "till": '2026-07-03',
    'interval': 60,
}

candles_response = requests.get(
    url=candles_url,
    params=candles_params,
    timeout=30
)

print(candles_response.url)
print(candles_response.status_code)

candles_data = candles_response.json()
candles_df = pd.DataFrame(
    data=candles_data["candles"]["data"],
    columns=candles_data["candles"]["columns"],
)

candles_df.head()
```

```python
#Получить свечи указанного инструмента по выбранному режиму торгов SBER - 1час
candles_sber_engine = "stock"
candles_sber_market = "shares"
candles_sber_board = "TQBR"
candles_sber_security = "sber"

candles_sber_url = rf'https://iss.moex.com/iss/engines/{candles_sber_engine}/markets/{candles_sber_market}/boards/{candles_sber_board}/securities/{candles_sber_security}/candles.json'
candles_sber_params = {
    "from": '2026-07-01',
    "till": '2026-07-03',
    'interval': 60,
}

candles_sber_response = requests.get(
    url=candles_sber_url,
    params=candles_sber_params,
    timeout=30
)

print(candles_sber_response.url)
print(candles_sber_response.status_code)

candles_sber_data = candles_sber_response.json()
candles_sber_df = pd.DataFrame(
    data=candles_sber_data["candles"]["data"],
    columns=candles_sber_data["candles"]["columns"],
)

candles_sber_df.head(15)
```

```python
#Дневная история SBER за период
history_sber_engine = "stock"
history_sber_market = "shares"
history_sber_board = "TQBR"
history_sber_security = "sber"

history_sber_url = rf'https://iss.moex.com/iss/history/engines/{history_sber_engine}/markets/{history_sber_market}/boards/{history_sber_board}/securities/{history_sber_security}.json'
history_sber_params = {
    "from": '2026-07-01',
    "till": '2026-07-15',
    "lang": 'ru',
}

history_sber_response = requests.get(
    url=history_sber_url,
    params=history_sber_params,
    timeout=30
)

print(history_sber_url)
print(history_sber_response.status_code)

history_sber_data = history_sber_response.json()


history_sber_df = pd.DataFrame(
    data=history_sber_data["history"]["data"],
    columns=history_sber_data["history"]["columns"],
)

history_sber_df.head(20)
```

```python
# Для выгрузки данных, если в первую партию выгрузилось не все.

cursor = dict(
    zip(
        history_sber_data['history.cursor']["columns"],
        history_sber_data['history.cursor']["data"][0],
    )
)

history_sber_params["start"] = cursor['PAGESIZE']

history_sber_response = requests.get(
    url=history_sber_url,
    params=history_sber_params,
    timeout=30,
)

history_sber_response.raise_for_status()

print(history_sber_response.url)

history_sber_data = history_sber_response.json()

history_sber_page_2_df = pd.DataFrame(
    data=history_sber_data["history"]["data"],
    columns=history_sber_data["history"]["columns"],
)

history_sber_page_2_df.head()
```
